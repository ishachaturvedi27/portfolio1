## Refactor the Cat-Petting Guide

Remove the custom CSS and classes, replace them with semantic HTML tags, and apply the appropriate OCS classes. Then press Run.

{% endcapture %}

{% capture code4 %}

// Refactor the HTML to use semantic tags and OCS classes.

outputElement.innerHTML = [
  '<div class="ocs_card">',
  '  <h2 class="ocs_section-title">How to Pat a Cat</h2>',
  '  <h3>Have you ever wanted to...</h3>',
  '  <p>pat a cat? Well good for you we will give step by step instructions on how to pat a cat!</p>',
  '  <h4><strong>How to pat a cat:</strong></h4>',
  '  <ol>',
  '    <li>put out your hand</li>',
  '    <li>touch the cat</li>',
  '    <li>move your hand back and forth</li>',
  '  </ol>',
  '  <h3>Congrats! You pat a cat :D</h3>',
  '</div>'
].join('');
