```python
# CODE_RUNNER: Popcorn Hack 1 - Read the API, Then Use It

import math
import random

# Read the API before calling a procedure
print("random.randint:", random.randint.__doc__)
print("math.ceil:", math.ceil.__doc__)
print(dir(math))
print(dir(random))

points_per_game = [24, 18, 31, 27, 22]

# random.choice picks one item from a list
print("Random game to review:", random.choice(points_per_game), "points")

# Changed the jersey range from 1-99 to 1-55
jersey = random.randint(1, 55)
print("Random jersey number:", jersey)

# Changed the court size from 144 to 225
side = math.sqrt(225)
print("Square court side:", side, "feet")

# math.ceil rounds up to the next whole number
average = sum(points_per_game) / len(points_per_game)
rounded_up = math.ceil(average)
print("Points per game:", average, "-> rounded up:", rounded_up)
```

```python
# CODE_RUNNER: Popcorn Hack 2 - Flask API
rom flask import Flask, jsonify

app = Flask(__name__)

@app.route("/score")
def score():
    return jsonify({"home": 58, "away": 52, "quarter": 3})

@app.route("/roster")
def roster():
    return jsonify({"players": ["Isha", "Emily", "Aashi"]})

@app.route("/coach")
def coach():
    return jsonify({"coach": "Coach Smith"})

client = app.test_client()

print("/score says:", client.get("/score").get_json())
print("/roster says:", client.get("/roster").get_json())
print("/coach says:", client.get("/coach").get_json())
```

```python
# CODE_RUNNER: Homework Hack - Season Stats Report

import json
import statistics
from datetime import date
import random

games = [
    {"date": "2026-09-20", "opponent": "Torrey Pines", "kills": 8},
    {"date": "2026-09-22", "opponent": "Scripps Ranch", "kills": 12},
    {"date": "2026-09-25", "opponent": "Del Norte", "kills": 10},
    {"date": "2026-09-28", "opponent": "Poway", "kills": 15},
]

def summarize_season(season):
    """
    Summarize a list of volleyball games.

    Parameters:
        season: list of dictionaries, each with "date", "opponent", and "kills"

    Returns:
        dictionary with total_kills, average_kills, best_game,
        and kills_by_opponent
    """
    kills = [game["kills"] for game in season]

    kills_by_opponent = {}
    for game in season:
        opponent = game["opponent"]
        kills_by_opponent[opponent] = kills_by_opponent.get(opponent, 0) + game["kills"]

    return {
        "total_kills": sum(kills),
        "average_kills": round(statistics.mean(kills), 1),
        "best_game": max(kills),
        "kills_by_opponent": kills_by_opponent,
    }

report = summarize_season(games)

# Calculate how many days are left until the next game
next_game = date(2026, 10, 10)
report["days_until_next_game"] = (next_game - date.today()).days

# Use random to demonstrate another standard library module
report["random_game_to_review"] = random.choice(games)["opponent"]

print(json.dumps(report, indent=2))

# requirements.txt lists external packages only.
# json, datetime, statistics, and random come with Python,
# so they do not need to be listed.
requirements_txt = """pandas
Flask
"""

print()
print("requirements.txt:")
print(requirements_txt)
```
