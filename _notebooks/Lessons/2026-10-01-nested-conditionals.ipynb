```python
# CODE_RUNNER: Popcorn Hack 1 - Test the After-School Homework Plan

is_weekday = True
homework_finished = False

if is_weekday:
    if homework_finished:
        print("Free time")
    else:
        print("Finish homework")
else:
    print("Weekend plan")
```

```python
# CODE_RUNNER: Popcorn Hack 2 - Repair the Student Account Check

logged_in = True
has_permission = False

if logged_in:
    print("Account found")
    if has_permission:
        print("Page opened")
    else:
        print("Permission needed")
else:
    print("Log in first")
```

```python
# CODE_RUNNER: Popcorn Hack 3 - Test the Lunch and Sports-Practice Plan

lunch_packed = True
practice_after_school = True

if lunch_packed:
    if practice_after_school:
        print("Bring lunch and an extra snack for practice")
    else:
        print("Bring lunch")
else:
    if practice_after_school:
        print("Buy lunch and bring an extra snack for practice")
    else:
        print("Buy lunch")
```

```python
# CODE_RUNNER: Homework Hack - Build a High School Daily Routine Dashboard

student = {
    "logged_in": True,
    "homework_finished": False,
    "lunch_packed": True,
    "practice_today": True
}

if student["logged_in"]:
    if student["homework_finished"]:
        if student["practice_today"]:
            if student["lunch_packed"]:
                print("Homework finished. Bring lunch and go to practice.")
            else:
                print("Homework finished. Buy lunch before practice.")
        else:
            if student["lunch_packed"]:
                print("Homework finished. Bring lunch and enjoy your afternoon.")
            else:
                print("Homework finished. Plan to get lunch and enjoy your afternoon.")
    else:
        if student["practice_today"]:
            print("Finish homework before practice.")
        else:
            print("Finish homework before your free time.")
else:
    print("Log in first before checking your daily schedule.")

# Test different combinations of the student values
```
