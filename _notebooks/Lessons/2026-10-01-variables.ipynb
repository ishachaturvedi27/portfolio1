## Tests

### Test 1: Bonus = 5

Prediction: The stored total will stay 40, and the recalculated total will become 45.

Actual output: 40, then 45.

Explanation: The stored total does not automatically change when level two stars change. It becomes 45 only after I recalculate it.

### Test 2: Bonus = 0

Prediction: All totals will be 40.

Actual output: 40.

Explanation: With no bonus added, the total remains 40.

```python
# CODE_RUNNER: UESL 3.1 Popcorn - Keep the checkpoint

stars = 8
checkpoint_stars = stars
stars = stars + 4

print(stars)
print(checkpoint_stars)
```

### Prediction 

Prediction: stars = 12, checkpoint_stars = 8.

Actual output: 12, 8.

Explanation: The checkpoint keeps the value 8 from before the stars were increased. Changing stars to 12 does not automatically change the stored checkpoint.

```python
# CODE_RUNNER: UESL 3.1 Homework - Build Your Star Counter

game_name = "Star Quest"
reduced_motion = True

level_one_stars = 12
level_two_stars = 18
level_three_stars = 10

total_stars = level_one_stars + level_two_stars + level_three_stars

print(game_name)
print(reduced_motion)
print(total_stars)

# Add a bonus of 5 to level two
level_two_stars = level_two_stars + 5

# The stored total does not change automatically
print(total_stars)

# Recalculate the total
total_stars = level_one_stars + level_two_stars + level_three_stars
print(total_stars)
```
